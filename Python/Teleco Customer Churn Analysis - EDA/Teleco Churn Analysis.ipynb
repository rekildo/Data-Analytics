```python
def conv(value):
    if value == 1:
        return "yes"
    else:
        return "no"

df['SeniorCitizen'] = df['SeniorCitizen'].apply(conv)
```

```python
ax = sns.countplot(x = 'Churn', data = df)
ax.bar_label(ax.containers[0])
plt.title('count of customers by churn')
plt.show()
```

```python
plt.figure(figsize = (3,4))
gb = df.groupby('Churn').agg({'Churn':'count'})
plt.pie(gb['Churn'], labels = gb.index, autopct = '%1.2f%%')
plt.title('percentage of churned customers', fontsize=10)
plt.show()
```

```python
plt.figure(figsize = (3,4))
sns.countplot(x = df['gender'], data = df, hue = 'Churn')
plt.title('churn by gender')
plt.show()
```

```python
plt.figure(figsize = (3,4))
ax = sns.countplot(x = df['SeniorCitizen'], data = df, hue = 'Churn')
ax.bar_label(ax.containers[0])
plt.title('churn by senior citizen')
plt.show()
```

```python

# Calculate percentages within each SeniorCitizen group
churn_pct = pd.crosstab(
    df['SeniorCitizen'],
    df['Churn'],
    normalize='index'
) * 100

# Plot stacked bar
ax = churn_pct.plot(
    kind='bar',
    stacked=True,
    figsize=(5, 4)
)

plt.title('Churn by Senior Citizen (%)')
plt.xlabel('Senior Citizen')
plt.ylabel('Percentage')
plt.xticks(rotation=0)
plt.legend(title='Churn')

# Add percentage labels
for container in ax.containers:
    labels = [
        f'{bar.get_height():.1f}%' if bar.get_height() > 3 else ''
        for bar in container
    ]
    ax.bar_label(container, labels=labels, label_type='center')

plt.tight_layout()
plt.show()
```

```python
#greater % of senior citizen category have churned out
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import math

# Remove customerID
cols = [col for col in df.columns if col != 'customerID']

# Number of columns in each row
n_cols = 3

# Number of rows required
n_rows = math.ceil(len(cols) / n_cols)

fig, axes = plt.subplots(n_rows, n_cols, figsize=(18, 5 * n_rows))

# Flatten axes so we can loop easily
axes = axes.flatten()

for i, col in enumerate(cols):

    if df[col].dtype == 'object':
        sns.countplot(data=df, x=col, ax=axes[i], hue = df['Churn'] )
        axes[i].tick_params(axis='x', rotation=45)

    else:
        sns.histplot(data=df, x=col, ax=axes[i], kde=True)

    axes[i].set_title(col)

# Remove empty subplots
for j in range(i + 1, len(axes)):
    fig.delaxes(axes[j])

plt.tight_layout()
plt.show()
```
